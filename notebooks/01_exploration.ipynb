```python
import pandas as pd
from pathlib import Path
```

```python
PROJECT_ROOT  = Path("..")
DATA_DIR = PROJECT_ROOT/"data"/"raw"/"twcs.csv"
pd.set_option('display.max_colwidth', 1000)
```

```python
df = pd.read_csv(DATA_DIR)
print(df.columns)
```

```python
company_reply = df['inbound'] == False
valuables = df[company_reply]['author_id'].value_counts()
print(valuables.head(50))
```

```python
user_tweets = df['inbound'] == True
airways_mask = df['text'].str.contains("@Delta|@AmericanAir|@British_Airways", case=False)
combined = user_tweets & airways_mask
print(df[combined]['text'].head(25))
print(df[combined].shape)
```

```python
dataset = df[combined].sample(n=6000, random_state=42)
dataset.sample(25)
```
